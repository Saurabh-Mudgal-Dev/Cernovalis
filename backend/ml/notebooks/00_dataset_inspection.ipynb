# Dataset Inspection

Initial, read-only inspection of the raw ISOT news datasets
(`True.csv` and `Fake.csv`) located in `backend/ml/data/raw/`.

**Scope:** load both datasets and report their structure — shapes, column
names, data types, missing value counts, sample rows, and whether both files
share the same schema.

```python
from pathlib import Path

import pandas as pd
```

```python
# Resolve the raw data directory
candidates = (Path("../data/raw"), Path("backend/ml/data/raw"))
data_dir = next((p for p in candidates if p.is_dir()), None)
if data_dir is None:
    raise FileNotFoundError(f"Could not locate backend/ml/data/raw/ from {Path.cwd()}")

true_path = data_dir / "True.csv"
fake_path = data_dir / "Fake.csv"
for path in (true_path, fake_path):
    if not path.is_file():
        raise FileNotFoundError(f"Missing dataset: {path.resolve()}")

print(f"Using raw data directory: {data_dir}")
```

## Load the raw datasets

```python
true_df = pd.read_csv(true_path)
fake_df = pd.read_csv(fake_path)
```

## Shapes

```python
pd.DataFrame({"True.csv": true_df.shape, "Fake.csv": fake_df.shape}, index=["rows", "columns"])
```

## Column names and data types

```python
print("True.csv columns:", true_df.columns.tolist())
print("Fake.csv columns:", fake_df.columns.tolist())
```

```python
pd.DataFrame({"True.csv": true_df.dtypes, "Fake.csv": fake_df.dtypes}).rename_axis("column")
```

## Missing values

```python
pd.DataFrame({"True.csv": true_df.isna().sum(), "Fake.csv": fake_df.isna().sum()}).rename_axis("column")
```
